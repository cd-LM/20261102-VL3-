# Umsatzanalyse: Berliner Einzelhandel

Wir betrachten eine **fiktive Buchhandlung in Berlin-Neukölln** und ihre Umsätze von Montag bis Freitag, 26. bis 30. Oktober 2026. Die Zahlen sind Beispieldaten und keine echten Geschäftsdaten.

In diesem Notebook erstellen wir eine Tabelle und berechnen Kennzahlen mit einfachem Python.

## 1. Tagesumsätze erfassen

Jeder Eintrag enthält ein Datum, einen Wochentag und den Umsatz in Euro. Die geschweiften Klammern `{}` bilden jeweils einen Datensatz.

```python
umsatzdaten = [
    {"datum": "26.10.2026", "wochentag": "Montag", "umsatz": 1245.80},
    {"datum": "27.10.2026", "wochentag": "Dienstag", "umsatz": 1098.50},
    {"datum": "28.10.2026", "wochentag": "Mittwoch", "umsatz": 1432.20},
    {"datum": "29.10.2026", "wochentag": "Donnerstag", "umsatz": 1680.00},
    {"datum": "30.10.2026", "wochentag": "Freitag", "umsatz": 2245.70}
]
```

## 2. Tabelle anzeigen

Die Schleife geht die fünf Datensätze nacheinander durch und gibt Datum, Wochentag und Umsatz als Tabelle aus.

```python
print(f"{'Datum':<12} | {'Wochentag':<12} | {'Umsatz (Euro)':>14}")
print("-" * 45)
for eintrag in umsatzdaten:
    print(f"{eintrag['datum']:<12} | {eintrag['wochentag']:<12} | {eintrag['umsatz']:>14.2f}")
```

## 3. Kennzahlen berechnen

Wir addieren die Umsätze, berechnen den Durchschnitt und merken uns beim Durchlaufen der Liste den umsatzstärksten und den umsatzschwächsten Tag.

```python
gesamt = 0
umsatzstaerkster_tag = umsatzdaten[0]
umsatzschwaechster_tag = umsatzdaten[0]

for eintrag in umsatzdaten:
    gesamt += eintrag["umsatz"]

    if eintrag["umsatz"] > umsatzstaerkster_tag["umsatz"]:
        umsatzstaerkster_tag = eintrag

    if eintrag["umsatz"] < umsatzschwaechster_tag["umsatz"]:
        umsatzschwaechster_tag = eintrag

durchschnitt = gesamt / len(umsatzdaten)

print(f"Wochenumsatz: {gesamt:.2f} Euro")
print(f"Durchschnitt pro Tag: {durchschnitt:.2f} Euro")
print(f"Stärkster Tag: {umsatzstaerkster_tag['wochentag']} mit {umsatzstaerkster_tag['umsatz']:.2f} Euro")
print(f"Schwächster Tag: {umsatzschwaechster_tag['wochentag']} mit {umsatzschwaechster_tag['umsatz']:.2f} Euro")
```

## Einordnung

Für diese Beispieldaten beträgt der Wochenumsatz **7.702,20 Euro** und der durchschnittliche Tagesumsatz **1.540,44 Euro**. Der Freitag ist mit **2.245,70 Euro** der stärkste Tag, der Dienstag mit **1.098,50 Euro** der schwächste.

Das sind nur fünf fiktive Tage: Daraus lässt sich noch kein allgemeiner Wochentrend ableiten. Umsatz ist außerdem nicht dasselbe wie Gewinn, denn Kosten sind hier nicht berücksichtigt.
