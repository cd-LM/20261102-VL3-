In diesem Notebook wertest du drei Tagesumsätze aus. Dabei lernst du, wie Text, Python-Code und Ergebnisse in einem Dokument zusammenkommen.

## Wofür eignet sich Jupyter?

Jupyter ist besonders praktisch, wenn du Code schrittweise ausprobieren und die Ergebnisse direkt daneben sehen möchtest. Es wird oft für Datenanalyse, Unterricht, Experimente und anschauliche Dokumentation verwendet.

Ein Notebook besteht aus **Zellen**. Markdown-Zellen enthalten Erklärungen wie diese. Code-Zellen enthalten ausführbaren Python-Code. Der **Kernel** ist der laufende Python-Prozess, der den Code ausführt und Variablen zwischen den Zellen speichert.

## 1. Beispieldaten

Wir betrachten die Umsätze eines kleinen Verkaufsstands an drei Tagen. Führe die folgende Code-Zelle aus, zum Beispiel mit der Schaltfläche **Ausführen** oder `Shift+Enter`.

```python
umsaetze = [120.50, 95.00, 143.75, 110.25, 165.15]
print(umsaetze)
```

## 2. Summe und Durchschnitt berechnen

Die nächste Zelle verwendet die Liste aus der vorherigen Zelle. `sum(...)` addiert die Werte; durch die Anzahl der Werte geteilt ergibt sich der Durchschnitt.

```python
gesamt = sum(umsaetze)
durchschnitt = gesamt / len(umsaetze)

print(f'Gesamtumsatz: {gesamt:.2f} Euro')
print(f'Durchschnitt pro Tag: {durchschnitt:.2f} Euro')
```

## Ergebnis und Ausprobieren

Die Ausgabe lautet: **Gesamtumsatz: 359.25 Euro** und **Durchschnitt pro Tag: 119.75 Euro**.

Du kannst die Zahlen in `umsaetze` ändern und die Zellen erneut ausführen. So kannst du eine Idee sofort testen, ohne ein ganzes Programm neu zu starten. Wichtig: Wenn du Zellen in anderer Reihenfolge ausführst, können Variablen fehlen oder ältere Werte enthalten. Mit **Kernel neu starten und alle Zellen ausführen** lässt sich der Ablauf sauber von oben beginnen.
